MNIST images have 784 pixels, so the ambient dimension is 784. Four different quantities are often called "the dimension" of such data, and they answer different questions.

- Ambient dimension: the number of coordinates used to write down a datapoint (784 here).
- Intrinsic dimension: the dimension of the set the data concentrate near. Neither PCA nor UMAP measures it.
- Latent dimension: the size of $\mathbf z$ in a model. It is a modeling choice that you set, not a property of the data.
- Embedding dimension: the dimension of a visualization, such as the 2 axes of a UMAP plot.

PCA explained variance shows how many orthogonal linear directions capture most of the variance. A curve that rises quickly shows that a few linear directions already capture most of the variance, so the data are close to a low-dimensional linear subspace. A curve that rises slowly does not rule out a low-dimensional curved set, because a curved set can need many linear directions to cover it.

```python
from sklearn.decomposition import PCA
from torchvision.datasets import MNIST
from torchvision import transforms
import matplotlib.pyplot as plt
import numpy as np
import torch

# Load MNIST
transform = transforms.ToTensor()
mnist = MNIST(root="./data", train=True, transform=transform, download=True)

X = torch.stack([img for img, _ in mnist]).view(-1, 28 * 28).numpy()

# Apply PCA
pca = PCA(n_components=100)
X_pca = pca.fit_transform(X)

# Plot explained variance
plt.plot(np.cumsum(pca.explained_variance_ratio_))
plt.xlabel("Number of Principal Components")
plt.ylabel("Cumulative Explained Variance")
plt.title("PCA: cumulative explained variance of MNIST")
plt.grid(True)
plt.show()
```

UMAP builds a neighbor graph in the 784-dimensional space and finds a 2D layout that keeps local neighborhoods together. The plot shows that digits of the same class sit near each other. Distances between clusters and cluster sizes in the layout are not reliable, and the picture is not evidence that the data lie on a manifold of dimension 2.

```python
import umap.umap_ as umap

# Reduce to 2D using UMAP
X_2d = umap.UMAP(n_components=2).fit_transform(X[:5000])
y = [label for _, label in mnist][:5000]

# Plot
plt.scatter(X_2d[:, 0], X_2d[:, 1], c=y, cmap="tab10", s=5)
plt.colorbar()
plt.title("MNIST Digits in UMAP Space")
plt.show()
```
