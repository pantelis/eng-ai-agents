## Development Environment Setup

The course containers are the default. On an Apple Silicon Mac they cannot reach the GPU, so for the AI and computer vision courses you may install natively instead. See [macOS Native](/aiml-common/resources/environment/macos-native).

### Windows users

1. Install [WSL2](https://learn.microsoft.com/en-us/windows/wsl/install).

2. Ensure that you also follow [this tutorial](https://learn.microsoft.com/en-us/windows/wsl/tutorials/wsl-vscode) to setup VSCode properly aka the VSCode can access the WSL2 filesystem and work with the [remote docker containers](https://learn.microsoft.com/en-us/windows/wsl/tutorials/wsl-containers).

3. If you have an NVIDIA GPU in your system, ensure you have [enabled it](https://learn.microsoft.com/en-us/windows/wsl/tutorials/gpu-compute).

### Ubuntu users

Install docker in your system and the VSCode docker and remote extensions.

### macOS users

Install docker as above if you want the container, or follow [macOS Native](/aiml-common/resources/environment/macos-native) to install directly and use the GPU through Metal. Robotics coursework still needs the container.

### All users

Clone the [class repository](https://github.com/pantelis/eng-ai-agents). Windows users must clone it on the WSL2 filesystem. Take a screenshot of the terminal where you cloned it.

If you are using a container, build and launch it inside your IDE, then run `make start` and screenshot the terminal showing the `(your virtual env)` prefix. If you are running natively, `uv sync` in the repository root is the equivalent step.

Then select your virtual environment as the kernel and run the cell below. Save the output of every cell before submitting.

```python
import torch

if torch.cuda.is_available():
    device = torch.device("cuda")
elif torch.backends.mps.is_available():
    device = torch.device("mps")
else:
    device = torch.device("cpu")

print(f"torch {torch.__version__} on {device}")

if device.type == "cuda":
    print(torch.cuda.get_device_name(0))
```
